##  Where do A, F, L, T and N come from

The hoist=globals() command typically injects the following core Text-Fabric API objects into your global scope:

A: The App object itself (this is the same object that is also assigned to your N1904 variable).

F: The Feature API, used to access data features of nodes (e.g., F.lex.v(word_node)).

L: The Linguistic (or Locality) API, used to navigate between nodes (e.g., L.u(word_node, 'sentence') to go up to the containing sentence).

T: The Text API, used to get text representations of nodes (e.g., T.text(sentence_node)).

N: The Node API, which provides functions for working with node ranges and iterating over nodes (e.g., N.walk() to iterate through all nodes).

This API deals with the basic TF data model: a graph of nodes and edges, annotated by features.

## Simple functions

### Display verses:

```python
def reference(node):
  """
  Returns a formatted reference string (e.g., 'Mark 1:1') 
  from any Text-Fabric node.
  """
  # T.sectionFromNode() returns a tuple like ('Mark', 1, 1)
  book, chapter, verse = T.sectionFromNode(node)
  
  # Format it using an f-string
  return f"{book} {chapter}:{verse}"
```

## Display the Mk 1:1 verse by reference

```python
verse_node = T.nodeFromSection(('Mark', 1, 1))

# Get the Greek text (like before)
print(f"{reference(verse_node)} {T.text(verse_node)}")
```

## Displaying verse using a format template

Text-Fabric's data design enables a very flexible representation of the corpus text ([see this section](https://centerblc.github.io/N1904/textformats.html)) Each representation follows a text format, implemented as a template that maps the format to individual 'features': text, but also punctuation, signs, etc.

![Alt text](https://centerblc.github.io/N1904/features/images/details_surface_features.png)

This text formats in this Text-Fabric database are identified by unique names, like 'text-orig-plain', and the label reflects what the related format is about, as these names follow a structured naming schema, consisting of a string of keywords separated by hyphens (-).

`what`-`how`-`fullness`

Not all combinations are possible:

| keyword | value | meaning |
| ------ | ----- | ------- |
| what		| text		| words as they belong to the text |
| what		| lex		| lexemes of the words |
| how		| orig		| the original Greek script (all Unicode) |
| how		| unaccent	| the original Greek script without accents |
| how		| translit	| transliteration into latin alphabeth |
| fullness	| full		| complete text with text-critical markers |
| fullness	| plain		| complete text without text-critical markers |

## Using Mk 1:1 as a structured node and display it our way 

```python
verse_node = T.nodeFromSection(('Mark', 1, 1))

if verse_node:
    # Get all 'word' nodes contained within the verse
    words = L.d(verse_node, 'word')
    # Print the verse itself
    print(f"\n{T.text(verse_node)}\n")
    
    # Print a formatted header
    print(f"{'Greek Text':<15} | {'Greek Lemma':<15} | {'English':<45} | {'Translation'}  ")
    print(f"{'--------------':<15} | {'--------------':<15} | {'---------------------------------------------':<45} | {'-----------------'}")

    # Loop through each word node and get its features
    for word in words:
        # Get the 'text' (Greek)
        greek = F.text.v(word)
        
        # Get the English 'gloss' for each Greek word, as per the Biblical Greek Vocabulary Builder (BGVB).
        gloss = F.gloss.v(word)

        # get the English translation based on the Berean Interlinear Bible
        trans = F.trans.v(word)

        lemma = F.lemma.v(word)
        
        # Print the formatted table row
        print(f"{greek:<15} | {lemma:<15} | {gloss:<45} | {trans} ")
```

## Looking for Ἰησοῦς in Mark 1

```python
verses = T.nodeFromSection(('Mark', 1))
# Get all 'word' nodes contained within the verse
words = L.d(verses, 'word')
lookingFor='Ἰησοῦς'
refs=[]
# FoundWordNodes = [word for word in words if F.lemma.v(word) == lookingFor]
for word in words:
        # Get the 'text' (Greek) and its lemma
        greek = F.text.v(word)
        lemma = F.lemma.v(word)
        # Get the English 'gloss' for each Greek word, as per the Biblical Greek Vocabulary Builder (BGVB).
        gloss = F.gloss.v(word)
        # get the English translation based on the Berean Interlinear Bible
        trans = F.trans.v(word)
        if F.lemma.v(word) == lookingFor:
            book, chapter, verse_number = T.sectionFromNode(word)
            parent_verse = T.nodeFromSection((book, chapter, verse_number))
            parent_verse_text = T.text(parent_verse)
            # Get all 'word' nodes contained within the verse
            parent_words = L.d(parent_verse, 'word')
            parent_translation = []
            for parent_word in parent_words:
                parent_translation.append(F.trans.v(parent_word))
            parent_translation_phrase = " ".join(parent_translation)
            refs.append(f'{book} {chapter}:{verse_number}\n {parent_verse_text}\n {parent_translation_phrase}\n')
            
print(f'\nFound references of "{lookingFor}": {len(refs)}\n')
for ref in refs:
    print(f'{ref}')
        
```
