```python
import spacy, requests, pandas, grc_odycy_joint_trf
from collections import Counter
```

```python
source = "data/Mark_Raw.txt"
file = open(source, 'r')
text = file.read()
```

```python
nlp = spacy.load("grc_odycy_joint_trf")
doc = nlp(text)
```

## `εὐαγγελίου` occurences in Mark

```python
target_lemma = "εὐαγγέλιον"
print(f"\n--- Finding '{target_lemma}' ---")

occurrences = []
for token in doc:
    if token.lemma_ == target_lemma:
        result = {
            "form": token.text,
            "lemma": token.lemma_,
            "context": token.sent.text.strip()
        }
        occurrences.append(result)

if occurrences:
    print(f"Found {len(occurrences)} occurrences of '{target_lemma}':\n")
    
    for i, occ in enumerate(occurrences, 1):
        print(f"Occurrency #{i}:")
        print(f"  > Exact form: {occ['form']}")
        print(f"  > Context: {occ['context']}")
        print("-" * 20)
else:
    print(f"No occurrence of '{target_lemma}' found.")
```
